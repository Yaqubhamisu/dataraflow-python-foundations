## 1. Classes and Objects

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_details(self):
        print(f"Brand: {self.brand}, Model: {self.model}")


# Instantiate two objects
car1 = Car("Toyota", "Camry")
car2 = Car("Tesla", "Model 3")

# Print their details
car1.display_details()
car2.display_details()
```

## 2. Class Variables

```python
class Counter:
    # Class variable to track total instances
    count = 0

    def __init__(self):
        # Increment the class variable each time a new object is created
        Counter.count += 1


# Create several objects
obj1 = Counter()
obj2 = Counter()
obj3 = Counter()
obj4 = Counter()

# Print total instances created
print("Total instances created:", Counter.count)
```

## 3. Inheritance and Method Overriding

```python
# Parent class
class Animal:
    def speak(self):
        print("Animal sound")


# Child class overriding speak()
class Dog(Animal):
    def speak(self):
        print("Woof!")


# Demonstration
animal = Animal()
animal.speak()  # Prints: Animal sound

dog = Dog()
dog.speak()  # Prints: Woof!
```

## 4. Multilevel Inheritance

```python
# Base class
class LivingBeing:
    is_alive = True


# Animal inherits from LivingBeing
class Animal(LivingBeing):
    pass


# Bird inherits from Animal
class Bird(Animal):
    def fly(self):
        print("Flying high!")


# Instantiate Bird and demonstrate inherited attributes/methods
my_bird = Bird()

# Inherited from LivingBeing
print("Is alive:", my_bird.is_alive)

# Method defined on Bird
my_bird.fly()
```

## 5. Iterators

```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.limit:
            raise StopIteration
        
        val = self.current
        self.current += 2
        return val


# Demonstration using a for loop (limit set to 10)
even_gen = EvenNumbers(10)
for num in even_gen:
    print(num)
```

## 6. Polymorphism

```python
class Cat:
    def sound(self):
        return "Meow"


class Dog:
    def sound(self):
        return "Woof"


# Function calling .sound() polymorphically
def animal_sound(animal):
    print(animal.sound())


# Demonstration
my_cat = Cat()
my_dog = Dog()

animal_sound(my_cat)  # Prints: Meow
animal_sound(my_dog)  # Prints: Woof
```

## 7. Global and Nonlocal Scope

```python
# 1. Global variable modified inside a function
counter = 0


def update_global():
    global counter
    counter += 10


update_global()
print("Global counter:", counter)  # Prints: 10


# 2. Nonlocal variable modified inside a nested function
def outer_function():
    message = "Initial Message"

    def inner_function():
        nonlocal message
        message = "Modified Message"

    inner_function()
    print("Nonlocal message:", message)


outer_function()  # Prints: Modified Message
```

## 8. Multiple Inheritance and Method Resolution Order

```python
class Flyer:
    def move(self):
        return "Flying"


class Swimmer:
    def move(self):
        return "Swimming"


# Duck inherits from both Flyer and Swimmer
class Duck(Flyer, Swimmer):
    def move(self):
        # Uses super() to call move() according to the Method Resolution Order (MRO)
        return super().move()


# Demonstration
duck = Duck()
print("Duck movement:", duck.move())  # Prints: Flying
print("MRO:", Duck.__mro__)
```

## 9. Fibonacci Iterator

```python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.a > self.n:
            raise StopIteration
        
        fib_number = self.a
        self.a, self.b = self.b, self.a + self.b
        return fib_number


# Demonstration
fib = Fibonacci(50)
for num in fib:
    print(num, end=" ")
```

## 10. Polymorphism with Different Shapes

```python
import math


class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)


class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


# Store instances in a list
shapes = [Rectangle(4, 5), Circle(3), Triangle(6, 4)]

# Calculate total area polymorphically without checking types
total_area = sum(shape.area() for shape in shapes)

print(f"Total Area: {total_area:.2f}")
```

# Featured Project: Library Management System

This project brings several OOP concepts together in a small library system.

### Features
- Different book types using inheritance
- Polymorphic `get_description()` methods
- Student and teacher members with different borrowing limits
- Iteration over books and borrowed books
- Protected and private attributes
- Book search by title or author
- Tracking of total borrowed books


```python
# 1. OOP & Inheritance + 2. Polymorphism

class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_borrowed = False

    def get_description(self):
        return f"'{self.title}' by {self.author} ({self.year})"


class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        base_desc = super().get_description()
        return f"[EBook] {base_desc} - {self.file_size}MB, Format: {self.format}"


class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        base_desc = super().get_description()
        return f"[PrintedBook] {base_desc} - {self.weight}kg, Dimensions: {self.dimensions}"


class Member:
    total_borrowed_books = 0  # Class-level variable across all members

    def __init__(self, name, member_id, borrow_limit):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []  # Protected attribute
        self.borrow_limit = borrow_limit

    def __iter__(self):
        # Iterator for Member to iterate over borrowed books
        return iter(self._borrowed_books)

    def borrow_book(self, book):
        if book.is_borrowed:
            print(f"Sorry, '{book.title}' is currently checked out.")
            return False
        if len(self._borrowed_books) >= self.borrow_limit:
            print(f"Borrowing limit reached for {self.name} (Limit: {self.borrow_limit}).")
            return False

        book.is_borrowed = True
        self._borrowed_books.append(book)
        Member.total_borrowed_books += 1
        print(f"{self.name} borrowed '{book.title}'.")
        return True


class StudentMember(Member):
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id, borrow_limit=2)  # Student limit = 2
        self.grade_level = grade_level


class TeacherMember(Member):
    def __init__(self, name, member_id, department):
        super().__init__(name, member_id, borrow_limit=5)  # Teacher limit = 5
        self.department = department


class Library:
    def __init__(self, library_name):
        self.__library_name = library_name  # Private attribute
        self.books = []
        self.members = []

    # Getter for private attribute __library_name
    @property
    def library_name(self):
        return self.__library_name

    # Setter for private attribute __library_name
    @library_name.setter
    def library_name(self, name):
        if name and isinstance(name, str):
            self.__library_name = name
        else:
            print("Invalid library name.")

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def __iter__(self):
        # Iterator for Library to iterate over all books
        return iter(self.books)

    def search_books(self, query):
        query_lower = query.lower()
        results = [
            book for book in self.books
            if query_lower in book.title.lower() or query_lower in book.author.lower()
        ]
        return results

    def display_available_books(self):
        print(f"\n Available Books in {self.__library_name} ")
        available = [b for b in self.books if not b.is_borrowed]
        if not available:
            print("No books currently available.")
        for book in available:
            print(book.get_description())



# Create Library
lib = Library("City Central Library")

# Instantiate Books
b1 = EBook("Python 101", "Guido van Rossum", 2020, 5, "PDF")
b2 = PrintedBook("Clean Code", "Robert C. Martin", 2008, 1.2, "20x15 cm")
b3 = PrintedBook("Design Patterns", "Erich Gamma", 1994, 1.5, "24x18 cm")

lib.add_book(b1)
lib.add_book(b2)
lib.add_book(b3)

# Instantiate Members
student = StudentMember("Alice", "S001", "Grade 12")
teacher = TeacherMember("Dr. Smith", "T100", "Computer Science")

lib.add_member(student)
lib.add_member(teacher)

# Demonstration: Borrowing & Limits (Polymorphism & Scope)
print("Borrowing Tests ")
student.borrow_book(b1)
student.borrow_book(b2)
student.borrow_book(b3)  # Exceeds student limit of 2

# Demonstration: Iterator on Library (Iterating all books)
print("\n Iterating over all books in Library ")
for book in lib:
    print(book.get_description())

# Demonstration: Iterator on Member (Iterating borrowed books)
print(f"\n Iterating over {student.name}'s borrowed books ")
for book in student:
    print(f"- {book.title}")

# Demonstration: Search Functionality
print("\n Search Results for 'clean' ")
search_results = lib.search_books("clean")
for book in search_results:
    print(book.get_description())

# Demonstration: Class-level borrowed count tracking
print(f"\nTotal books borrowed across all members: {Member.total_borrowed_books}")

```
