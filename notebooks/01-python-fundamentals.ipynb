```python
def my_calc(number1, number2, operation):
    if operation == "+":
        return number1 + number2
    elif operation == "-":
        return number1 - number2
    elif operation == "*":
        return number1 * number2
    else:
        return "Invalid operation"
print(my_calc(5, 3, "+"))
print(my_calc(5, 3, "-"))
print(my_calc(5, 3, "*"))
```

## 2. Practical Python Exercises

```python
# Student Grade System
marks = {"English": 75, "Maths": 79, "Physics": 80, "Statistics": 70, "Computer Science": 94}
total_marks = marks["English"] + marks["Maths"] + marks["Physics"] + marks["Statistics"] + marks["Computer Science"]
average = total_marks / 5
if average >= 80:
    Grade = "A"
elif average >=60:
    Grade = "B"
elif average >= 40:
    Grade = "C"
else:
    Grade = "F"
print("Grade:", Grade)
```

```python
# Simple Calculator
def calculator(num1, num2):
    return num1 + num2, num1 - num2, num1 * num2, num1 / num2
print("Select an operation: +, -, *, /")
operation = input("Operation: ")
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
if operation == "+":
    print("The result is " , calculator(num1, num2)[0])
elif operation == "-":
    print("The result is " , calculator(num1, num2)[1])
elif operation == "*":
    print("The result is " , calculator(num1, num2)[2])
elif operation == "/":
    print("The result is " , calculator(num1, num2)[3])
else:
    print("Invalid operation")
```

```python
# Shopping Cart
cart = {}
num_of_items = int(input("How many items do you want to add to your cart? "))
for i in range(num_of_items):
    item_name = input("Enter item name: ")
    price = float(input("Enter item price ($): "))
    cart[item_name] = price
print("\n--- YOUR SHOPPING CART ---")
print(cart)

total_cost = 0

for price in cart.values():
    total_cost = total_cost + price
print("Total Cost: $", total_cost)
```

```python
# Guessing Game
import random
secret_number = random.randint(1, 20)
user_guess = int(input("Enter your guess (1-20): "))
print("Secret number was:", secret_number)
print("Your guess was:", user_guess)

if user_guess > secret_number:
    print("Hint: Too high!")
elif user_guess < secret_number:
    print("Hint: Too low!")
else:
    print("Congratulations! You guessed it right!")
```

```python
# File Handling
students = []
print("Enter student names (type 'done' when you are finished):")
while True:
    name = input("Enter student name: ")
    if name.lower() == "done":
      break    
    students.append(name)
file = open("students.txt", "w")
for name in students:
    file.write(name + "\n")
file.close()
file = open("students.txt", "r")
content = file.read()
file.close()

print("\n--- STUDENTS SAVED IN FILE ---")
print(content)
```
